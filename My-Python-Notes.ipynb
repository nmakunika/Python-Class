```python
import pandas as pd

df = pd.DataFrame({
    "aridity_index": [0.10, 0.30, 0.50, 0.70],
    "ideal_children": [2, 4, 3, 5]
})

def classify_climate(aridity):
    if aridity < 0.20:
        return "Arid"
    elif aridity < 0.65:
        return "Semi-arid"
    else:
        return "Humid"

df["climate_category"] = df["aridity_index"].apply(classify_climate)

print(df)
```

```python
# Demographic and climate risk classification

age = 32
ideal_children = 4
aridity_index = 0.35
temperature = 25.5
precipitation = 650

if age >= 18 and ideal_children <= 2:
    fertility_category = "Low"

elif age >= 18 and ideal_children <= 4:
    fertility_category = "Moderate"

elif age >= 18 and ideal_children > 4:
    fertility_category = "High"

else:
    fertility_category = "Not eligible for adult classification"


# Climate classification
if aridity_index < 0.20:
    climate_category = "Arid"

elif aridity_index >= 0.20 and aridity_index < 0.65:
    climate_category = "Semi-arid"

else:
    climate_category = "Humid"


# Climate risk classification
if temperature >= 30 or precipitation < 400:
    climate_risk = "High"

elif temperature >= 25 and precipitation < 700:
    climate_risk = "Moderate"

else:
    climate_risk = "Low"


# Overall classification
if climate_risk == "High" and fertility_category == "High":
    overall_risk = "High demographic-climate vulnerability"

elif climate_risk == "Moderate" or fertility_category == "Moderate":
    overall_risk = "Moderate demographic-climate vulnerability"

else:
    overall_risk = "Low demographic-climate vulnerability"


# Display results
print("Age:", age)
print("Ideal number of children:", ideal_children)
print("Aridity Index:", aridity_index)
print("Temperature:", temperature)
print("Precipitation:", precipitation)

print("\nFertility category:", fertility_category)
print("Climate category:", climate_category)
print("Climate risk:", climate_risk)
print("Overall assessment:", overall_risk)
```
